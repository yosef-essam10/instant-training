# Kayfa AI Agent - Synthetic Course Data Generator
This notebook uses Faker to generate a realistic course catalog for the Kayfa AI Agent project and saves it to `data/courses.csv`.

```python
import random
import pandas as pd
from faker import Faker

fake = Faker()
random.seed(42)
Faker.seed(42)
```

```python
TRACKS = [
    "Artificial Intelligence",
    "Data Science",
    "SOC",
    "Cybersecurity",
    "Frontend",
    "Backend",
    "Fullstack",
    "Cloud Computing",
    "Mobile Development",
    "UI/UX Design",
]

LEVELS = ["Beginner", "Intermediate", "Advanced"]
LANGUAGES = ["Arabic", "English", "Arabic & English"]

COURSE_NAME_TEMPLATES = {
    "Artificial Intelligence": ["Intro to Machine Learning", "Deep Learning Fundamentals", "AI for Business", "NLP with Python", "Computer Vision Bootcamp"],
    "Data Science": ["Data Analysis with Python", "Data Science Diploma", "SQL for Data Science", "Data Visualization Masterclass", "Statistics for Data Science"],
    "SOC": ["SOC Analyst Diploma", "SOC Fundamentals", "Threat Detection and Response", "SIEM Tools in Practice", "Incident Handling Essentials"],
    "Cybersecurity": ["Ethical Hacking Basics", "Network Security Diploma", "Penetration Testing 101", "Cybersecurity Awareness", "Malware Analysis Intro"],
    "Frontend": ["HTML CSS JavaScript Bootcamp", "React for Beginners", "Frontend Diploma", "Responsive Web Design", "Vue.js Essentials"],
    "Backend": ["Node.js Backend Diploma", "Python Backend Development", "APIs with FastAPI", "Backend Fundamentals", "Django for Beginners"],
    "Fullstack": ["Fullstack Web Development Diploma", "MERN Stack Bootcamp", "Fullstack with Django and React", "Fullstack Diploma Advanced", "Fullstack Career Path"],
    "Cloud Computing": ["AWS Fundamentals", "Cloud Practitioner Diploma", "DevOps and Cloud Basics", "Azure for Beginners", "Cloud Architecture Essentials"],
    "Mobile Development": ["Flutter Bootcamp", "Android Development Diploma", "iOS Development with Swift", "React Native Essentials", "Mobile Apps from Scratch"],
    "UI/UX Design": ["UI UX Design Diploma", "Figma for Beginners", "User Research Fundamentals", "Design Systems 101", "Product Design Bootcamp"],
}
```

```python
def generate_courses():
    rows = []
    for track in TRACKS:
        for name in COURSE_NAME_TEMPLATES[track]:
            is_free = random.random() < 0.15
            price = 0 if is_free else random.choice([499, 799, 999, 1499, 1999, 2499, 2999])
            level = random.choice(LEVELS)
            duration_weeks = random.randint(2, 16)
            language = random.choice(LANGUAGES)
            description = fake.paragraph(nb_sentences=3)
            rows.append({
                "track": track,
                "course_name": name,
                "level": level,
                "price": price,
                "duration": f"{duration_weeks} weeks",
                "language": language,
                "is_free": is_free,
                "description": description,
            })
    return pd.DataFrame(rows)

df = generate_courses()
df.head(10)
```

```python
import os

output_path = os.path.join("data", "courses.csv")
os.makedirs("data", exist_ok=True)
df.to_csv(output_path, index=False)
print(f"Saved {len(df)} courses to {output_path}")
```
